```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('loss_output.csv')
```

```python
# Extract the columns
epochs = data['Step']
loss = data['Loss']
```

```python
# Plot the data
plt.figure(figsize=(10, 8))
plt.plot(epochs, loss, marker='o', markersize = 3)
plt.title('Steps vs Loss')
plt.xlabel('Steps')
plt.ylabel('Loss')
plt.grid(True)

# Set y-axis intervals of 0.1
y_min, y_max = loss.min(), loss.max()
plt.yticks(np.arange(y_min, y_max + 0.1, 0.1))

plt.show()
```
